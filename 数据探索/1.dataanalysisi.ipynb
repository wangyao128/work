```python
%%sql

-- SELECT * from all_trading_days limit 10
select a.*,b.total_market_cap
from cn_stock_real_bar1d a
inner join cn_stock_valuation b on a.date = b.date and a.instrument = b.instrument
where  a.date = '2026-02-12' and b.total_market_cap < 10000000000
order by b.total_market_cap DESC
```
