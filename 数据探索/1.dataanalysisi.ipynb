```python
#导入股票池
import pandas as pd
from datetime import date

# 读取 GB2312 编码的 CSV 文件
df = pd.read_csv('/home/aiuser/work/我的文档/量化交易/股票池/自选股_20260422.csv', encoding='utf-8')
#print(df.head())


df.columns = [str(col).strip() for col in df.columns]

# 解析股票代码
stocks = []

# 定义后缀映射关系
suffix_map = {'SZ': '.SZ', 'SH': '.SH', 'BJ': '.BJ'}

for _, row in df.iterrows():
    code_raw = str(row.get('股票代码', '')).strip()
    name = str(row.get('股票名称', '')).strip()
    type = str(row.get('分类标签', '')).strip()
    # 清理代码中的空格和控制字符
    # 遍历映射关系，查找并替换
    for prefix, suffix in suffix_map.items():
        if code_raw.startswith(prefix):
            # 如果是前缀开头 (如 SZ300666)，去掉前缀并加上后缀
            code = code_raw[len(prefix):] + suffix
            #print(f"Matched prefix: {prefix}, code: {code}")
            break
        elif code_raw.isdigit():
        # 北交所代码规则判断：
        # 920开头 (最新号段)
        # 8 开头 (如 83, 87)
        # 4 开头 (如 430)
        # 92 开头 (部分老代码)
            if (code_raw.startswith('920') or 
                code_raw.startswith('8') or 
                code_raw.startswith('4') or 
                code_raw.startswith('92')):
                code = code_raw + '.BJ'
            else:
                code = code_raw 
        else:
            code = code_raw.upper()

    stocks.append({'code': code, 'name': name,'type': type})

print(len(stocks))
print(stocks)

import dai 
df = pd.DataFrame(stocks)
today = date.today()
df['date'] = today.strftime('%Y-%m-%d')
# 可选：定义分区，可以用于数据访问加速
df[dai.DEFAULT_PARTITION_FIELD] = df["date"]

dai.DataSource.write_bdb(
    data=df,
    # datasource id是全局唯一的，支持小写字母、数字、下划线，以字母开始
    id="stock_pool_wyc",
    # 数据插入时，根据unique_together如果有重复的，会去重.如果有分区,则需要传入索引参数indexes
    unique_together=["date", "code"],
    indexes=["date"],
)
```

```python
%%sql
-- 1. 先清理可能存在的旧临时表（防止重复创建报错）
DROP TABLE IF EXISTS temp_stock_base;
DROP TABLE IF EXISTS temp_stock_factor;
DROP TABLE IF EXISTS temp_stock_result;

-- 2. 第一步：获取基础数据
-- 筛选出时间范围内、符合状态要求的股票数据
CREATE TEMP TABLE temp_stock_base AS
SELECT 
    a.instrument,
    a.date,
    a.close,
    a.adjust_factor
FROM cn_stock_bar1d a
INNER JOIN cn_stock_status b ON a.instrument = b.instrument
WHERE a.date BETWEEN '2020-01-01' AND '2026-04-17' 
  AND b.st_status = 0 
  AND b.suspended = 0 
  AND b.is_risk_warning = 0 
  AND b.date = '2026-04-17';

-- 3. 第二步：计算关键因子
-- 为每只股票提取“最新日期”的复权因子，作为全量计算的基准
CREATE TEMP TABLE temp_stock_factor AS
SELECT 
    instrument,
    MAX(CASE WHEN date = '2026-04-17' THEN adjust_factor END) AS latest_factor
FROM temp_stock_base
GROUP BY instrument;

-- 4. 第三步：最终计算与输出
-- 关联因子表，计算前复权最高/最低，并格式化输出
CREATE TEMP TABLE temp_stock_result AS
SELECT 
    t1.instrument,
    -- 计算前复权最高/最低价：原始极值 / 最新因子
    MAX(t1.close) / t2.latest_factor AS max_close,
    MIN(t1.close) / t2.latest_factor AS min_close,
    -- 计算最新收盘价的前复权值
    MAX(CASE WHEN t1.date = '2026-04-17' THEN t1.close END) / t2.latest_factor AS latest_close
FROM temp_stock_base t1
JOIN temp_stock_factor t2 ON t1.instrument = t2.instrument
GROUP BY t1.instrument, t2.latest_factor;

-- 5. 查询最终结果
SELECT 
    instrument, 
    max_close, 
    min_close, 
    latest_close,
    ROUND(latest_close/ NULLIF(max_close, 0), 2) as ratio_pct,
    ROUND(latest_close/ NULLIF(min_close, 0), 2) as dpp
FROM temp_stock_result
where ratio_pct > 0.3
ORDER BY ratio_pct asc;


```
